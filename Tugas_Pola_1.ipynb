```python
import numpy as np
import matplotlib.pyplot as plt

# Membaca data TSV
traffic = np.genfromtxt("web_traffic.tsv", delimiter='\t')

# Menampilkan 10 data pertama
print("10 data pertama:")
print(traffic[:10])

# Menampilkan ukuran data
print("\nUkuran data:")
print(traffic.shape)

# Memisahkan kolom
x = traffic[:, 0]
y = traffic[:, 1]

# Menghapus data yang memiliki NaN
x = x[~np.isnan(y)]
y = y[~np.isnan(y)]

# Membuat scatter plot
plt.scatter(x, y)

plt.title("Web Traffic Last Month")
plt.xlabel("Time")
plt.ylabel("Hits/hour")

# Menampilkan label minggu
plt.xticks(
    [w * 7 * 24 for w in range(10)],
    ['week %i' % w for w in range(10)]
)

plt.autoscale(tight=True)
plt.grid()

plt.show()
```
